```python
Q1 = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outlier = df[(df["Quantity"] < lower) | (df["Quantity"] > upper)]

print("Batas bawah:", lower)
print("Batas atas:", upper)
print("Jumlah outlier:", len(outlier))
```

```python
import pandas as pd

# Ringkasan audit
ringkasan = pd.DataFrame({
    "Pemeriksaan": [
        "Jumlah Data",
        "Jumlah Kolom",
        "Jumlah Duplicate",
        "Jumlah Outlier Quantity"
    ],
    "Hasil": [
        df.shape[0],
        df.shape[1],
        df.duplicated().sum(),
        len(outlier)
    ]
})

# Missing value per kolom
missing_value = df.isnull().sum().reset_index()
missing_value.columns = ["Kolom", "Jumlah Missing"]

# Tipe data
tipe_data = df.dtypes.reset_index()
tipe_data.columns = ["Kolom", "Tipe Data"]

# Permasalahan yang ditemukan
permasalahan = pd.DataFrame({
    "Jenis Permasalahan": [
        "Missing Value",
        "Duplicate Data",
        "Data Tidak Valid",
        "Outlier"
    ],
    "Hasil Audit": [
        "8 kolom memiliki 1.000 missing value",
        "Tidak ditemukan duplicate baris",
        "Tidak ditemukan nilai Quantity 0 atau negatif",
        "Tidak ditemukan outlier pada Quantity"
    ]
})

# Simpan ke Excel dengan beberapa sheet
with pd.ExcelWriter("Initial Data Audit.xlsx", engine="openpyxl") as writer:
    ringkasan.to_excel(writer, sheet_name="Ringkasan Audit", index=False)
    missing_value.to_excel(writer, sheet_name="Missing Value", index=False)
    tipe_data.to_excel(writer, sheet_name="Tipe Data", index=False)
    permasalahan.to_excel(writer, sheet_name="Permasalahan", index=False)

print("Initial Data Audit.xlsx berhasil dibuat.")
```
